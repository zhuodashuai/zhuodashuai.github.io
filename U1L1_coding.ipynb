## 3. Uploading data downloaded from Canvas

The next cell looks for a file named **`diamonds.csv`** in the current runtime. If the file is absent in Colab, a file chooser will appear. Select the copy of `diamonds.csv` that you downloaded from Canvas.

The check makes the cell safe to run more than once: it asks for an upload only when the required file is missing. The same notebook also works in local Jupyter when `diamonds.csv` is placed in the same folder as the notebook.

```python
DATA_FILE = Path("diamonds.csv")

if not DATA_FILE.exists():
    try:
        from google.colab import files
    except ImportError as exc:
        raise FileNotFoundError(
            f"Place {DATA_FILE.name} in the same folder as this notebook."
        ) from exc

    print(f"Please select the Canvas file named exactly: {DATA_FILE.name}")
    uploaded = files.upload()

    if DATA_FILE.name not in uploaded:
        uploaded_names = ", ".join(uploaded) if uploaded else "no files"
        raise FileNotFoundError(
            f"Expected {DATA_FILE.name}, but received: {uploaded_names}. "
            "Rename or re-download the Canvas file and run this cell again."
        )

if not DATA_FILE.exists():
    raise FileNotFoundError(f"Could not find {DATA_FILE.name}.")

print(f"Ready to read: {DATA_FILE.resolve()}")
```

## 4. Reading and checking the data

A CSV file stores a rectangular table as plain text. `pd.read_csv()` reads that table into a pandas `DataFrame`. We immediately check its dimensions and column names so that a missing, renamed, or incorrect file fails early.

```python
diamonds = pd.read_csv(DATA_FILE)

expected_columns = [
    "carat", "cut", "color", "clarity", "depth",
    "table", "price", "x", "y", "z"
]

assert diamonds.shape == (53_940, 10), (
    f"Unexpected shape {diamonds.shape}; expected (53940, 10)."
)
assert diamonds.columns.tolist() == expected_columns, (
    "The column names do not match the course data file."
)

print("PASS: diamonds.csv was uploaded and read successfully.")
print(f"Rows: {diamonds.shape[0]:,}")
print(f"Columns: {diamonds.shape[1]}")
display(diamonds.head())
```

```python
print("Column names:")
print(diamonds.columns.tolist())

print("\nData types inferred from the CSV file:")
display(diamonds.dtypes.rename("dtype").to_frame())

print("Missing values in each column:")
display(diamonds.isna().sum().rename("missing").to_frame())
```

## 5. Including a plot

The full dataset contains 53,940 observations. For this first preview, we draw a reproducible random sample of 1,000 rows so the plot is quick and uncluttered. The analysis below is only a workflow demonstration; we will learn how to select and interpret visualizations in Lecture 2.

```python
diamonds_preview = diamonds.sample(n=1_000, random_state=4710)

point_color = "steelblue"  # Try another named color.
plot_title = "Diamond price and carat: a first look"  # Edit the title.

fig, ax = plt.subplots(figsize=(6, 4))
sns.scatterplot(
    data=diamonds_preview,
    x="carat",
    y="price",
    color=point_color,
    alpha=0.55,
    s=24,
    ax=ax,
)
ax.set(
    title=plot_title,
    xlabel="Carat",
    ylabel="Price (US dollars)",
)
plt.show()
```

> Using Python, NumPy, seaborn, and matplotlib, simulate 10,000 observations from a standard normal distribution with random seed 4710. Plot a density-scaled histogram and overlay the theoretical standard normal density curve. Include clear labels and explain each line.

Before running suggested code, check:

The following is one checked implementation.

```python
rng = np.random.default_rng(4710)
normal_sample = rng.normal(loc=0, scale=1, size=10_000)

x_grid = np.linspace(-4, 4, 400)
normal_density = np.exp(-(x_grid**2) / 2) / np.sqrt(2 * np.pi)

number_of_bins = 50  # Try 20 or 100 and compare the result.
histogram_color = "lightblue"

fig, ax = plt.subplots(figsize=(6, 4))
sns.histplot(
    normal_sample,
    bins=number_of_bins,
    stat="density",
    color=histogram_color,
    edgecolor="white",
    ax=ax,
)
ax.plot(x_grid, normal_density, color="red", linewidth=2)
ax.set(
    title="Simulated standard normal distribution",
    xlabel="Value",
    ylabel="Density",
)
plt.show()
```
